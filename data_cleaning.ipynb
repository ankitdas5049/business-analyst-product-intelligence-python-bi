```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)
```

```python
df.isnull().sum()
missing = (
    df.isnull()
      .sum()
      .to_frame("Missing Values")
)

missing["Percentage"] = (
    missing["Missing Values"] / len(df) * 100
)

missing.sort_values("Percentage", ascending=False)
```

```python
# Removing duplicate products
df.duplicated().sum()
```

```python
# Checking ASIN duplicates
df["asin"].duplicated().sum()
# If exists
df = df.drop_duplicates(subset="asin")
```
